# 本人清唱 A/B 试听
选择 RVC Comparison 内核。Shift+Enter 逐格运行；先评分，再运行最后一格揭晓。这里播放保存的 WAV，不需要 GPU。

```python
from pathlib import Path
from IPython.display import Audio,display,FileLink
import json
root=Path.cwd()
result=root / "runs/self_rvc_compare"
if not (result / "comparison.json").exists():
    candidates=sorted((root / "runs").glob("*/comparison.json"),key=lambda p:p.stat().st_mtime)
    if not candidates: raise RuntimeError("尚未生成结果；先完成云端实验。")
    result=candidates[-1].parent
print("结果目录：",result)
for name in ["listening/input.wav","blind/X.wav","blind/Y.wav"]:
    path=result/name
    print(name)
    if path.exists(): display(Audio(filename=str(path),normalize=False))
    else: print("尚未生成",name)
display(FileLink(str(result/"listening_scores.csv")))
```
