```python
from pathlib import Path
from IPython.display import Audio, FileLink, display
lab_root = Path.cwd()
if not (lab_root / "voicelab").exists():
    raise RuntimeError("请在 voice-lab 目录打开这个 notebook")
wav_files = sorted((lab_root / "runs").rglob("*.wav"))
print("找到", len(wav_files), "个音频文件")
for path in wav_files:
    print(path.relative_to(lab_root))
```

```python
for path in wav_files:
    print(str(path.relative_to(lab_root)))
    display(Audio(filename=str(path)))
    display(FileLink(str(path.relative_to(lab_root))))
```
